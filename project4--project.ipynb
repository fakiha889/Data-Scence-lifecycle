# **TASK 1 : customer churn analysis**
# **DESCRIPTION**
**This lab translates the six core stages of the Data Science Lifecycle.We will demonstrate each step of the structured data science lifecycle using a telecom customer churn scenario.**

# **STEP 1 : SETTING THE GOAL**


```python
project_charter = {
 'Research_Goal': 'Predict telecom customer churn to reduce annual subscriber loss',
 'Business_Benefit': 'Identify at-risk accounts early; targeted retention saves $250k/year',
 'Required_Data': 'CRM customer profiles, monthly billing logs, customer support tickets',
 'Timetable': '4 Weeks (Week 1: Prep, Week 2: EDA, Week 3: Model, Week 4: Deployment)',
 'Deliverables': 'Automated daily churn risk alert report & executive dashboard'
}
print('=== PROJECT CHARTER ESTABLISHED ===')
for key, val in project_charter.items():
  print(f"{key} : {val}")
```

# **STEP 2 : RETREVING DATA**

```python
# LOAD DATA
import pandas as pd
raw_crm_data = {
 'CustomerID': [101, 102, 103, 104, 105, 106],
 'MonthlySpend': [85.0, None, 45.0, 120.0, -25.0, 95.0],
 'TenureMonths': [12, 3, None, 24, 1, 18]
}
df_raw = pd.DataFrame(raw_crm_data)
print('Raw Data Shape:', df_raw.shape)
print('Missing Values Check:\n', df_raw.isnull().sum())
```

# **SREP 3 : DATA PREPARATION (cleaning,integration,transformation)**

```python
# Stage 3a: Cleansing (Fixing missing/negative entries)

df_clean = df_raw[df_raw['MonthlySpend'] >= 0].copy()
df_clean['MonthlySpend'].fillna(df_clean['MonthlySpend'].mean())
df_clean['TenureMonths'].fillna(df_clean['TenureMonths'].median())

## Stage 3b: Integration (Merging with Customer Tickets)

tickets_data = {'CustomerID': [101, 103, 104, 106], 'SupportTickets': [1, 5, 0, 2]}
df_tickets = pd.DataFrame(tickets_data)
df_merged = pd.merge(df_clean, df_tickets, on='CustomerID', how='left').fillna(0)

## Stage 3c: Transformation (Feature Engineering)

df_merged['Risk_Index'] = df_merged['SupportTickets'] / (df_merged['TenureMonths'] + 1)
print('Prepared Dataset:\n', df_merged)
```

# **STEP 4 : DATA EXPLORATION(EDA)**

```python
print('Descriptive Summary:\n', df_merged.describe())
correlation = df_merged['MonthlySpend'].corr(df_merged['SupportTickets'])
print(f'Spend vs Support Tickets Correlation: {correlation:.2f}')
```

# **STEP 5 : DATA MODELING / MODEL BUILDING**

```python
def predict_churn_risk(row):
 if row['SupportTickets'] >= 3 or row['Risk_Index'] > 0.5:
     return 'High Churn Risk'
     return 'Low Risk'
df_merged['Churn_Prediction'] = df_merged.apply(predict_churn_risk, axis=1)
print('Model Predictions:\n', df_merged[['CustomerID', 'Churn_Prediction']])
```

# **STEP 6 : PRESENTATION AND AUTOMATION**

```python
high_risk = df_merged[df_merged['Churn_Prediction'] == 'High Churn Risk']
print('=== AUTOMATED DAILY ACTION ALERT ===')
for idx, r in high_risk.iterrows():
 print(f'� ALERT: Customer {int(r["CustomerID"])} flagged! Send retention offer.')
```

# **STEP 2 : RETREVING THE DATA**

```python
import pandas as pd

hospital_data = {
    "Patient_ID": ["P101", "P102", "P103", "P104", "P105", "P106", "P107", "P108"],
    "Age": [65, 24, 48, 72, 35, None, 19, 61],
    "Emergency_Level": ["High", "Low", "Medium", "High", "Low", "Medium", "High", "Medium"],
    "Arrival_Time": ["09:00", "09:05", "09:10", None, "09:20", "09:25", "09:30", "09:35"],
    "Waiting_Time_Min": [10, 45, None, 25, 60, 40, 5, 35],
    "Department": ["Cardiology", "General", "Orthopedics", "Cardiology", "General", "Orthopedics", "Emergency", "General"]
}
df = pd.DataFrame(hospital_data)
print(df.shape)
print(df.isnull().sum())
```

# **STEP 3 : DATA PREPARATION (cleaning,integration,transformation)**

```python
# step 3a : cleaning(missing/negative values)
import pandas as pd

hospital_data = {
    "Patient_ID": ["P101", "P102", "P103", "P104", "P105", "P106", "P107", "P108"],
    "Age": [65, 24, 48, 72, 35, None, 19, 61],
    "Emergency_Level": ["High", "Low", "Medium", "High", "Low", "Medium", "High", "Medium"],
    "Arrival_Time": ["09:00", "09:05", "09:10", None, "09:20", "09:25", "09:30", "09:35"],
    "Waiting_Time_Min": [10, 45, None, 25, 60, 40, 5, 35],
    "Department": ["Cardiology", "General", "Orthopedics", "Cardiology", "General", "Orthopedics", "Emergency", "General"]
}
df = pd.DataFrame(hospital_data)
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Waiting_Time_Min"] = df["Waiting_Time_Min"].fillna(df["Waiting_Time_Min"].median())
df["Arrival_Time"] = df["Arrival_Time"].ffill()
print(df)

# step 3b : integration(combine data)
# Create DataFrames by selecting lists of columns instead of tuples
patient_data = df[["Patient_ID", "Age", "Emergency_Level"]]
waiting_time = df[["Patient_ID", "Waiting_Time_Min", "Department"]]

df_merged = pd.merge(patient_data, waiting_time, on="Patient_ID")
print(df_merged)

#step 3c : transformation(make data suitable for analysis)
def check_waiting_time(minutes):
  if minutes<=15:
    return "short"
  elif minutes<=30:
    return "moderate"
  else:
    return "too long"
df["waiting_cetagory"] = df["Waiting_Time_Min"].apply(check_waiting_time)
print(df)
```

# **STEP 4 : DATA EXPLORATION**

```python
print('Descriptive Summary:\n', df_merged.describe())

# Map qualitative Emergency_Level to numeric values to calculate correlation
emergency_mapping = {'Low': 1, 'Medium': 2, 'High': 3}
numeric_emergency = df_merged['Emergency_Level'].map(emergency_mapping)

correlation = df_merged['Waiting_Time_Min'].corr(numeric_emergency)
print(f'waiting time vs emergency level Correlation: {correlation:.2f}')
```

# **STEP 5 : DATA MODELING / DATA BUILDING**

```python
import pandas as pd

hospital_data = {
    "Patient_ID": ["P101", "P102", "P103", "P104", "P105", "P106", "P107", "P108"],
    "Age": [65, 24, 48, 72, 35, 48.0, 19, 61],
    "Emergency_Level": ["High", "Low", "Medium", "High", "Low", "Medium", "High", "Medium"],
    "Arrival_Time": ["09:00", "09:05", "09:10", "09:10", "09:20", "09:25", "09:30", "09:35"],
    "Waiting_Time_Min": [10, 45, 35.0, 25, 60, 40, 5, 35],
    "Department": ["Cardiology", "General", "Orthopedics", "Cardiology", "General", "Orthopedics", "Emergency", "General"]
}
df = pd.DataFrame(hospital_data)

def predict_waiting_time_risk(row):
    if row['Emergency_Level'] == 'High' and row['Waiting_Time_Min'] > 15:
        return 'High Risk'
    elif row['Waiting_Time_Min'] > 30:
        return 'High Risk'
    else:
        return 'Low Risk'
df['Waiting_Time_Prediction'] = df.apply(predict_waiting_time_risk, axis=1)
print('Model Predictions:\n', df[['Patient_ID', 'Waiting_Time_Prediction']])
```

# **STEP 6 : PRESENTATION AND AUTOMATION**

```python
high_risk = df[df['Waiting_Time_Prediction'] == 'High Risk']

print('=== AUTOMATED HOSPITAL ALERT SYSTEM ===')

for idx, r in high_risk.iterrows():
    print(f'ALERT: Patient {r["Patient_ID"]} flagged! Review waiting time and prioritize care according to clinical urgency.')
```
